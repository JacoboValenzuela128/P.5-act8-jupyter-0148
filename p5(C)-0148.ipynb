```python
datos = {
    "Producto": ["Bolsa_utiles", "Lapiceros", "Borradores", "Marcadores", "Marcatextos"],
    "Precio": [50, 15, 10, 70, 20],
    "Cantidad": [1, 2, 3, 7, 1]
}

# Data Frame o Marco de datos
df = pd.DataFrame(datos)
df["Total"] = df["Precio"] * df["Cantidad"]
print("-=Mostrar datos=-")
print("=== VENTAS DE PAPELERÍA ===")
print(df)
```
