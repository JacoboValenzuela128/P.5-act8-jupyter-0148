```python
datos = {
    "Dia": ["Lunes", "Martes", "Miércoles", "Jueves", "Viernes"],
    "Temperatura": [15, 20, 11, 25, 30]
}

# Data Frame o Marco de datos
df = pd.DataFrame(datos)
print("-=Mostrar datos=-")
print("=== TEMPERATURAS ===")
print(df)
```

```python
fila = df["Temperatura"].idxmax()

print("Día con mayor temperatura:", df.loc[fila, "Dia"])
print("Temperatura máxima:", df["Temperatura"].max(), "°C")
print("-------------------Jupyter Valenzuela--------------------------")
print("-------------------Valenzuela Arath NC:0148--------------------------")
```
